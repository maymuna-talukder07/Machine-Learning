```python
import pandas as pd

data = pd.read_csv("/content/UsaHousing/USA_Housing.csv")
data.head()
```

```python
data.info()
```

```python
data.head()
```

```python
data.drop(
    columns=['Address'],
    inplace=True
)
data.head()
```

```python
data['Area Population'] = data['Area Population'].astype('int64')
data.head()
```

```python
data.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(data=data)
```

```python
sns.jointplot(data=data , x='Avg. Area House Age' , y='Area Population' , kind='hex')
```

```python
corr = data.corr()
```

```python
sns.heatmap(corr, annot=True)
```

```python
import sklearn
```

```python
from sklearn.model_selection import train_test_split
Features = data.drop(columns="Price")
```

```python
Features
```

```python
target =data["Price"]
target
```

```python
X_train, X_test, y_train, y_test = train_test_split(Features, target , random_state=101 , train_size=.8)
```

```python
X_test
```

```python
X_train
```

```python
y_train
```

```python
from sklearn.linear_model import LinearRegression

model =LinearRegression()
```

```python
model.fit(X_train , y_train)
```

```python
model.coef_
```

```python
pd.DataFrame(
    index = X_train.columns,
    data = model.coef_,
    columns = ['coefficents']
)
```

```python
model.intercept_
```

```python
model.predict(X_test)
```

```python
sns.scatterplot(x = y_test , y = model.predict(X_test))
```

```python
sns.histplot(
    data=(y_test - model.predict(X_test)),
    kde=True
)
```

```python
from sklearn.metrics import mean_absolute_error , mean_squared_error
prediction = model.predict(X_test)
MAE = mean_absolute_error(y_test , prediction)
MAE
```

```python
mse = mean_squared_error(y_test , prediction)
mse
```

```python
import numpy as np

rmse = np.sqrt(mse)
rmse
```
